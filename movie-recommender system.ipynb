```python
movies = movies[['movie_id', 'title', 'overview','genres', 'keywords','cast','crew']]
```

```python
import ast
```

```python
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['genres']
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies['keywords']
```

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies['crew'][1]
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
        
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies['crew']
```

```python
movies.sample(5)
```

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)

```

```python
movies.head()
```

```python
movies['overview'] = movies['overview'].fillna('')

movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['crew']
```

```python
new = movies.drop(columns=['overview','genres','keywords','cast','crew'])
```

```python
new.head()
```

```python
import ast
import numpy as np

def safe_convert_tags(x):
    # If x is None or NaN (float), return empty list
    if x is None or (isinstance(x, float) and np.isnan(x)):
        return []
    
    # If x is already a list, keep it
    elif isinstance(x, list):
        return x
    
    # If x is a string representing list of dicts
    elif isinstance(x, str):
        try:
            parsed = ast.literal_eval(x)
            return [i['name'] for i in parsed]
        except:
            return []
    
    # Anything else, return empty list
    else:
        return []

# Apply to the column
new['tags'] = new['tags'].apply(safe_convert_tags)

# Keep only first 3 tags
# movies['tags'] = movies['tags'].apply(lambda x: x[:3])

# Remove spaces in multi-word tags
new['tags'] = new['tags'].apply(lambda x: [i.replace(" ", "") for i in x])

# Join tags into a single string for CountVectorizer
new['tags'] = new['tags'].apply(lambda x: " ".join(x))

print(new['tags'].head())
```

```python
new
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector = cv.fit_transform(new['tags']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
new[new['title'] == 'The Lego Movie'].index[0]
```

```python

def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
```

```python
recommend('Gandhi')
```
